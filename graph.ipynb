```python
for i in df.columns:
    if df[i].dtype == "str":
        continue
    plt.stem(df[i][:100], df["EstimatedSalary"][:100])
    plt.title(f"Relation between {i} VS Estimated Salary", color = "red", size = 30)
    plt.xlabel(i, color = "blue", size = 25)
    plt.ylabel("Estimated Salary", color = "green", size = 25)
    plt.show()
```
